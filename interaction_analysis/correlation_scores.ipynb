# Correlations across different evaluation metrics

````python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import argparse
import sys
import os
from pathlib import Path
from cltl.dialogue_evaluation.utils.constants import GRAPH_METRICS, LIKELIHOOD_METRICS, HUMAN_METRICS
from cltl.dialogue_evaluation.metrics_correlations import Correlator
````

The ```Correlator``` class in the ```cltl.dialogue_evaluation``` takes three files from an *EMISSOR* scenario as input and creates a matrix to calculate the correlation between each metrics.

Currently, you can get the correlations between the group A, B, and C graph metrics, two encoder likelihood scores and 8 human evaluation scores using the metrics from the [DSTC-9 competition](https://dstc9.dstc.community). 

The metrics are defined here per group:

https://github.com/leolani/cltl-dialogueevaluation/blob/main/src/cltl/dialogue_evaluation/utils/constants.py

You can create your custom list of metrics for comparison and the custom list to the paramter ```metrics``` below.
In the next example, we use all the 50 metrics.

The results are saved in the evaluation folder of a scenario, both as a correlation heatmap and as a csv file with the scores.

For a further explanation oif the scores and correlations see:

```@inproceedings{santamaria2022evaluating,
  title={Evaluating Agent Interactions Through Episodic Knowledge Graphs},
  author={Santamar{\'\i}a, Selene B{\'a}ez and Vossen, Piek and Baier, Thomas},
  booktitle={Proceedings of the 1st Workshop on Customized Chat Grounding Persona and Knowledge},
  pages={15--28},
  uri={https://aclanthology.org/2022.ccgpk-1.3/},
  year={2022}
}
```

````python
correlator = Correlator()
emissor_path = Path("./emissor")
scenario = "d5a6bc60-c19b-4c08-aee5-b4dd1c65c64d"
scenario_path = os.path.join(emissor_path, scenario)
print(scenario_path)
metrics=GRAPH_METRICS + LIKELIHOOD_METRICS + HUMAN_METRICS
graph_evalation = "graph_evaluation.csv"
llh_evalation="likelihood_evaluation_USR_context300.csv"
manual_evaluation="d5a6bc60-c19b-4c08-aee5-b4dd1c65c64d_manual_evaluation.csv"

correlator.correlate_metrics_single_scenario(scenario=Path(scenario_path), 
                                             graph_file=graph_evalation, 
                                             llh_file=llh_evalation, 
                                             manual_file = manual_evaluation, 
                                             metrics=metrics)
````
