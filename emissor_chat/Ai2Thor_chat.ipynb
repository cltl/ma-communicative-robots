# Chat in an Ai2Thor space with EMISSOR

In this notebook, we demonstrate how you can chat with the agent in the [Ai2Thor space](
https://ai2thor.allenai.org/ithor/documentation/environment-state) and save the so-called ```signals``` in EMISSOR.
For this, we will import two clients:

1) an Ai2ThorClient for instructing the agent in the space using actions and obtaining information about the space
2) a LeolaniChatClient so that we can 1) start a new scenario, 2) capture the interaction with the Ai2ThorClient as signals and 3) save the interaction on disk in the folder that is created for EMISSOR scenario.
  
There are three modalities of signals that take place during the ineraction:

1. text modality for the communication between the user and the agent
2. action modality for the actions caried out by the agent in the space (mostly navigation)
3. image modality for the objects found by the agent in the space

EMISSOR stores signals for each modality in a scenario folder with a unique name. At the start of an interaction, the new scenario folder is created together with a JSON file with the same name, the so-called scenario JSON. The scenario JSON contains the meta data for the scenario as a whole, including a temporal ruler in terms of a start and end time. The temporal ruler is used to align the signal in a temporal sequence.


**Reference**: Santamaría, Selene Báez, Thomas Baier, Taewoon Kim, Lea Krause, Jaap Kruijt, and Piek Vossen. "EMISSOR: A platform for capturing multimodal interactions as Episodic Memories and Interpretations with Situated Scenario-based Ontological References." In Proceedings of the 1st Workshop on Multimodal Semantic Representations (MMSR), pp. 56-77. 2021.


````python
from leolani_client import LeolaniChatClient, Action
emissor_path = "./emissor"
HUMAN="Piek"
AGENT="Ai2Thor"
leolaniClient = LeolaniChatClient(emissor_path=emissor_path, agent=AGENT, human=HUMAN)
````

````python
import ai2thor
from ai2thor.controller import Controller
import numpy as np
````

````python
from ai2thor_client import Ai2ThorClient
````

````python
ai2ThorClient = Ai2ThorClient()

max_context=50
AI = "AI"

utterance = "Hi %s. Tell me what to do." % HUMAN
print(AGENT+">"+utterance)
leolaniClient._add_utterance(AGENT, utterance) 

utterance = "This is what I can do:"+str(ai2ThorClient.what_i_can_do())
print(AGENT+">"+utterance)
leolaniClient._add_utterance(AGENT, utterance) 

utterance = input(HUMAN+"> ")
leolaniClient._add_utterance(HUMAN, utterance) 

while not (utterance.lower() == "stop" or utterance.lower() == "bye"):
        ai2ThorClient.process_instruction(utterance)
        for utterance in ai2ThorClient._answers:
            print(AGENT+">"+str(utterance))
            leolaniClient._add_utterance(AGENT, utterance)
            
        for obj, objectType, coord, image in ai2ThorClient._perceptions:
            leolaniClient._add_image(obj['name'], objectType, coord, image) 

        for action in ai2ThorClient._actions:
            leolaniClient._add_action(action)
            
        utterance = input(HUMAN+"> ")
        leolaniClient._add_utterance(HUMAN, utterance) 

ai2ThorClient._controller.stop()
##### After completion, we save the scenario in the defined emissor folder.
leolaniClient._save_scenario() 
````

## Modalities in EMISSOR

For each modality a separate JSON file is created: "text.json", "audio.json" and "image.json". Each JSON file lists the signals recorded in a modality with their time-stamps and possible annotations. In the current application, we will only deal with text and image singals. In the case of text signals, we annotate each signal with the source of each turn, which, in this application, can be the HUMAN, the AGENT or ACTION. The source ACTION is a special case, because EMISSOR does not support actions as a separate modality. Therefore, any action that is carried out by the agent in AI2Thor is stored as a text signal with the source "ACTION".

The ```text.json``` thus contain a mixture of turns and actions.

## Loading an EMISSOR scenario and retrieving the signals

After the interaction is saved in EMISSOR it can be loaded again and each signal can be retrieved. The next code shows how to do this for a specific scenario and to print the text signals and image signals.

We first load the EMISSOR data and instantiate a controller.

````python
from emissor.persistence import ScenarioStorage
from emissor.representation.scenario import Modality

emissor_path = "emissor"
scenario = "0985799c-80eb-416c-b55e-f11ffc4ac259"

scenario_storage = ScenarioStorage(emissor_path)
scenario_ctrl = scenario_storage.load_scenario(scenario)
````

Using the controller, we can pull out all the text signal and print all the properties for each:

````python
text_signals = scenario_ctrl.get_signals(Modality.TEXT)
for text_signal in text_signals:
    print(text_signal)
    break
````

We can do the same for the image signals:

````python
image_signals =  scenario_ctrl.get_signals(Modality.IMAGE)
for image_signal in image_signals:
    print(text_signal)
````
