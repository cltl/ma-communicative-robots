```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
model = SentenceTransformer('sentence-transformers/roberta-large-nli-stsb-mean-tokens')
```

```python
sentences = ["The image shows a kitchen interior. It features wooden cabinets, marble countertops, a toaster, and a bread loaf on the counter. There is also a shelf against the far wall with various items, and a door in the background. The floor appears to be wooden, and the room is well-lit.", "A kitchen with a stove and sink"]

embeddings = model.encode(sentences)
print(embeddings)
```

```python
# Calculate cosine similarity
similarity = cosine_similarity(embeddings[0].reshape(1,-1), embeddings[1].reshape(1,-1))
print(similarity)
```
