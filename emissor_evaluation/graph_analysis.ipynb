# Episodic Knowledge Graph analysis

In this notebook, we demonstrate how interactions can be scores on a turn-by-turn basis by the triples extracted from the utterances.
These triples capture all kind of properties of the interaction: the content communicated, the perspective of the source but also the implication of communicating the content or the perspective such as does it raise a conflict or uncertainty (Baez et el., 2022).

These triples reflect properties of each turn but also cumulate into Knowledge Graph Patterns over the sequence of exchange. The resulting Knowledge Graph can be described in terms of three ghroups of properties:

* Mathematical properties (Group A): abstract properties of the graph independent of the semantics of the nodes and the edges, e.g. density of the graph
* Semantic properties (Group B): volume of classes, properties, and relationships
* Interaction properties (Group C): 

**Reference:**

Báez Santamaría,, S., Vossen, P., & Baier, T. (2022). Evaluating agent interactions through episodic knowledge graphs. In Proceedings of the 1st Workshop on Customized Chat Grounding Persona and Knowledge@ COLING2022 (pp. 15-28). Association for Computational Linguistics (ACL). arXiv e-prints (2022): arXiv-2209
