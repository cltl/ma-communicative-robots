## Spreadsheet for manual evaluation

This notebook generates a spreadsheet from an EMISSOR scenario to manually evaluate the conversation.
The turns are represented row-by-row with a unique identifier for the signal, the speaker and the actual utterance.
Additional columns are give for a reference response and any evaluation metrics that are provided.

The metrics added through this notebook are based on the Ninth Dialog System Technology Challeng: [DSTC9-track3](https://dstc9.dstc.community).
A PDF with the expl;anation of the evaluation criteria is given in this evaluation repository: DSCT9-track3-Annotation Questionairre.pdf.

For the course, you need to generate the spreadsheet for a selected scenario and annote 50 system responses.
You need to score each metrics on a scale from 1 (lowest, bad) to 5 (highest, flawless). Please only use integers for your score within this range.
There is also a column for the Overall Human Rating. Please score this independently of the specific metrics. This can be different from the average of the specific metrics. Ask a fellow student from your group to evaluate the same selection of turns (without showing your ratings).

Calculate the interannotator agreement across both annotations, as explained in class. For this you can use a standard scores such as ```Krippendorff’s α```, ```Spearman```, ```Pearson```, or ```Mean Absolute Error```, as provided in scikit-learn or pandas.

````python
import os
import pandas as pd
from emissor.persistence import ScenarioStorage
from emissor.representation.scenario import Modality
from emissor.representation.scenario import Signal, TextSignal
import emissor_util as util
````

````python
def evaluate_conversation(emissor:str, scenario_id:str, metrics=None):
    ### Create the scenario folder, the json files and a scenarioStorage and scenario in memory

    text_signals = util.get_text_signals_from_a_scenario(emissor,scenario_id)

    print('EMSISSOR:', emissor)
    print('Nr of turns:', len(text_signals), ' extracted from scenario: ', scenario_id)

    rows = []
    for text_signal in text_signals:
        turn_id = text_signal.id
        speaker = util.get_speaker_from_text_signal(text_signal)
        text = text_signal.text
        row = {"Turn": turn_id, "Speaker": speaker, "Response": text, "Reference Response": ""}
        for metric in metrics:
            row.update({metric:''})
        rows.append(row)
    df = pd.DataFrame(rows)
    # Save
    evaluation_folder = os.path.join(emissor, scenario_id, 'evaluation')
    if not os.path.exists(evaluation_folder):
        os.mkdir(evaluation_folder)
    file_name =  scenario_id+"_manual_evaluation.csv"
    file = os.path.join(evaluation_folder, file_name)
    df.to_csv(file, sep=";", index=False)
````

````python
EMISSOR="../emissor"
SCENARIO="2ad0c8b9-4e62-4f11-809c-57bf76487f95"
metrics = ["Overall Human Rating", "Engaging", "Specific", "Relevant", "Correct", "Understandable", "Coherent"]
evaluate_conversation(emissor=EMISSOR, scenario_id=SCENARIO, metrics=metrics)
````

The spreadsheet can be found in the ```evaluation``` subfolder of the scenario.
